```python
import numpy as np
import scipy.sparse as sparse
import matplotlib.pyplot as plt
from cg import cg
```

```python
n = 1000
d = 0.5 + np.sqrt(np.arange(1, n + 1))
A = sparse.diags([np.ones(n - 100),
                  np.ones(n - 1),
                  d,
                  np.ones(n - 1),
                  np.ones(n - 100)],
                 [-100, -1, 0, 1, 100], format='csr')
b = np.ones(n)
```

```python
res = []
x, _ = cg(A, b, tol=1e-16, maxiter=40, residuals=res)
```

In this experiment we will look at both the residual and the error of a calculation using a direct solver.


First set up a problem:

```python
import scipy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(556)
n = 10
A = np.random.rand(n,n)
xstar = np.ones((n,))
b = A @ xstar
for i in range(3):
    A = A @ A
    b = A @ xstar
```

```python
print("Cond(A)=",np.linalg.cond(A))
```

```python
nla = np.linalg
```

```python
x = np.linalg.solve(A, b)
```

```python
print("||r|| = ", nla.norm(b-A @ x))
print("||e|| = ", nla.norm(xstar-x))
```

Here we can see that since the condition number is high, a small residual does not lead to a small error.

```python
errs = []
conds = []
ress = []
xstar = np.ones(2)
for eps in np.logspace(-1,-14, 100):
    
    A = np.array([[1, 0], [0, eps]])
    V,_ = np.linalg.qr(np.random.randn(2,2))
    A = V.T @ A @ V
    b = A.dot(xstar)
    
    x = np.linalg.solve(A, b)
    
    diff = np.max(np.abs(x-xstar))   # error
    cond = np.linalg.cond(A)         # condition
    res = np.max(np.abs(b - A @ x))  # residual
    
    errs.append(diff)
    conds.append(cond)
    ress.append(res)
    print(f'{eps:.4e} {cond:.4e} {diff:.4e}, {res:.4e}')
```

```python
plt.loglog(conds, errs, 'bo')
plt.loglog(conds, ress, 'rs')
```

```python
import numpy as np
import scipy.linalg
import scipy as sp
import matplotlib.pyplot as plt
from common import set_figure
%matplotlib inline
```

```python
n = 64
h = 1/n

arow = np.zeros(n-1)
arow[:2] = [2, -1]
A = (1/h**2) * sp.linalg.toeplitz(arow)
omega = 2/3
Dinv = np.diag(1/np.diag(A))
```

```python
np.random.seed(7729292)
ustar = np.random.rand(n-1)
f = A @ ustar
u0 = np.random.rand(n-1)
```

```python
set_figure(heightratio=0.6)
fig, ax = plt.subplots()
x = np.linspace(0, 1, n+1)
e = np.zeros_like(x)
u = u0.copy()
e[1:-1] = ustar - u
ax.plot(x, e, '-', color='0.4', label=f'0', lw=0.5, clip_on=False)
plist = [2,5,20]
clist = ['tab:blue', 'tab:green', 'tab:red']
c = 0
for i in range(1, 21):
    if i in plist:
        e[1:-1] = ustar - u
        ax.plot(x, e, '-', color=clist[c], label=f'{i}', lw=2)
        c += 1
    u[:] = u + omega * Dinv @ (f - A @ u)
ax.set_ylim([-.6,.6])
ax.set_xlabel('$x$')
ax.set_ylabel('error $e_i(x)$')
ax.legend(bbox_to_anchor=(1.04,1),frameon=False)
plt.savefig('jacobi-smooth.pdf', bbox_inches='tight')
```

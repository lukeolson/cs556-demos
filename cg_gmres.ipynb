# CG and GMRES convergence

The test matrices are $A = S B S^{-1}$, where $S$ is upper bidiagonal with ones on the
diagonal and $\beta$ on the superdiagonal, and $B$ holds the eigenvalues
$\lambda_1, \ldots, \lambda_n$ (or a Jordan block).  The right-hand side $b$ is
chosen so that the solution of $Ax = b$ is $x = S\mathbf{1}$, which gives
$b = S B \mathbf{1}$.  Start from $x_0 = 0$.

```python
import numpy as np
import scipy.linalg as la
import matplotlib.pyplot as plt
```

```python
def make_problem(B, beta):
    """Build A = S B S^{-1}, b, and the exact solution x = S 1.

    B: n x n array
    beta: float
    """
    n = B.shape[0]
    S = np.eye(n) + beta * np.eye(n, k=1)
    A = la.solve(S.T, (S @ B).T).T     # S B S^{-1} without forming S^{-1}
    x = S @ np.ones(n)
    b = S @ (B @ np.ones(n))
    return A, b, x
```

```python
# example: beta = 0.9, lambda_j = j, n = 200
n = 200
B = np.diag(np.arange(1, n + 1, dtype=float))
A, b, x = make_problem(B, beta=0.9)
print(la.norm(A @ x - b) / la.norm(b))
```
